```python
data3 = pd.DataFrame({
    'value': [2, 45, -23, 85, 28, 2, 35, -12]
})
data3
```

```python
data3["scaled"] = (data3["value"] - data3["value"].min()) / (data3["value"].max() - data3["value"].min())
data3
```

```python
data3["normalized"] = (data3["value"] - data3["value"].mean()) / data3["value"].std()

print(f"Průměr: {data3['normalized'].mean()}")
print(f"Rozptyl: {data3['normalized'].var()}")
```
